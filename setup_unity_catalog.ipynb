```python
%sql
-- Skapa catalog
CREATE CATALOG IF NOT EXISTS marathos_catalog;

-- Använd catalogen
USE CATALOG marathos_catalog;

-- Skapa schema för raw data
CREATE SCHEMA IF NOT EXISTS marathon_raw;

-- Skapa volume för att ladda upp filer
CREATE VOLUME IF NOT EXISTS marathon_raw.marathon_volume;

-- Skapa schema för Bronze, Silver och Gold (för pipelinen)
CREATE SCHEMA IF NOT EXISTS marathos_catalog.marathon_bronze;
CREATE SCHEMA IF NOT EXISTS marathos_catalog.marathon_silver;
CREATE SCHEMA IF NOT EXISTS marathos_catalog.marathon_gold;

-- Verifiera att alla scheman är skapade
SHOW SCHEMAS IN marathos_catalog;
```

```python
print("Bronze:", spark.table("marathos_catalog.marathon_bronze.bronze_races").count())
print("Silver:", spark.table("marathos_catalog.marathon_silver.silver_cleaned").count())
print("dim_event:", spark.table("marathos_catalog.marathon_gold.dim_event").count())
print("fct_results:", spark.table("marathos_catalog.marathon_gold.fct_results").count())
```

```python
%sql
DROP TABLE IF EXISTS marathos_catalog.marathon_silver.silver_cleaned;
DROP TABLE IF EXISTS marathos_catalog.marathon_gold.dim_athlete;
DROP TABLE IF EXISTS marathos_catalog.marathon_gold.dim_event;
DROP TABLE IF EXISTS marathos_catalog.marathon_gold.fct_results;
DROP MATERIALIZED VIEW IF EXISTS marathos_catalog.marathon_gold.vw_50km_races;
DROP MATERIALIZED VIEW IF EXISTS marathos_catalog.marathon_gold.vw_100km_races;
```

```python
%sql
CREATE OR REPLACE VIEW marathos_catalog.marathon_gold.vw_yearly_participants AS
SELECT event_year, COUNT(*) as antal_deltagare
FROM marathos_catalog.marathon_silver.silver_cleaned
GROUP BY event_year
```

```python
%sql
DROP MATERIALIZED VIEW IF EXISTS marathos_catalog.marathon_gold.vw_50km_races;
```

```python
%sql
DROP TABLE IF EXISTS marathos_catalog.marathon_gold.dim_athlete;
DROP TABLE IF EXISTS marathos_catalog.marathon_gold.dim_event;
DROP TABLE IF EXISTS marathos_catalog.marathon_gold.fct_results;
DROP MATERIALIZED VIEW IF EXISTS marathos_catalog.marathon_gold.vw_50km_races;
DROP MATERIALIZED VIEW IF EXISTS marathos_catalog.marathon_gold.vw_100km_races;
```

```python
%sql
CREATE OR REPLACE VIEW marathos_catalog.marathon_gold.vw_50km_clean AS
SELECT DISTINCT
    event_name,
    athlete_id,
    event_id,
    athlete_performance
FROM marathos_catalog.marathon_silver.silver_cleaned
WHERE event_distance = '50km'
```
