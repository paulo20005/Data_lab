```python
%sql
-- Skapa catalog
CREATE CATALOG IF NOT EXISTS marathos_catalog;

-- Använd catalogen
USE CATALOG marathos_catalog;

-- Skapa schema för raw data
CREATE SCHEMA IF NOT EXISTS marathon_raw;

-- Skapa volume för att ladda upp filer
CREATE VOLUME IF NOT EXISTS marathon_raw.marathon_volume;

-- Skapa schema för Bronze, Silver och Gold (för pipelinen)
CREATE SCHEMA IF NOT EXISTS marathos_catalog.marathon_bronze;
CREATE SCHEMA IF NOT EXISTS marathos_catalog.marathon_silver;
CREATE SCHEMA IF NOT EXISTS marathos_catalog.marathon_gold;

-- Verifiera att alla scheman är skapade
SHOW SCHEMAS IN marathos_catalog;
```

```python
%sql
DROP TABLE IF EXISTS marathos_catalog.marathon_bronze.bronze_races;
DROP TABLE IF EXISTS marathos_catalog.marathon_silver.silver_cleaned;
DROP TABLE IF EXISTS marathos_catalog.marathon_gold.dim_event;
DROP TABLE IF EXISTS marathos_catalog.marathon_gold.fct_results;
```
