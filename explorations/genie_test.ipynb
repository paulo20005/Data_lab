### Fördelar med Genie

1. **Man får svar snabbt** – Stakeholders behöver inte vänta på att en dataanalytiker ska hinna kolla. Man ställer bara frågan direkt och får svar på några sekunder.

2. **Mindre jobb för oss dataanalytiker** – När stakeholders själva kan kolla enkla grejer som "hur många sprang 50km?" slipper vi avbrytas hela tiden. Vi kan fokusera på svårare grejer istället.

3. **Lätt att använda** – Man behöver inte kunna SQL. Man skriver bara på vanlig svenska eller engelska, typ "visa topp 10 länder". Det är ganska smidigt.

4. **Alltid tillgängligt** – Genie fungerar dygnet runt. Stakeholders kan ställa frågor när de vill, även mitt i natten eller på helgen.

### Begränsningar

1. **Bara enkla frågor** – Genie är bra på att räkna och visa topplistor, men mer avancerade frågor blir det lätt fel på.

2. **Krångliga frågor funkar dåligt** – När man behöver göra flera joinar eller avancerad logik är det bättre att köra SQL själv. Genie blir lätt förvirrad.

3. **Beror på ren data** – Om datan inte är ren blir Genies svar fel. Man måste se till att underliggande data är bra annars funkar inte Genie som den ska.

### Rekommendation

Genie funkar bra för enkla standardfrågor som:

- Hur många deltagare var det på 50km?
- Vilket land hade flest löpare?
- Vilken är medeltiden för 100km?

-- Men om det är en viktig fråga som påverkar beslut bör man alltid dubbelkolla med manuell SQL. Då är man säker på att svaret  stämmer.



```python
%sql
-- Vilken distans är populärast?
SELECT e.event_distance, COUNT(DISTINCT CONCAT(f.athlete_id, f.event_id)) as antal_deltagare
FROM marathos_catalog.marathon_gold.fct_results f
JOIN marathos_catalog.marathon_gold.dim_event e ON f.event_id = e.event_id
GROUP BY e.event_distance
ORDER BY antal_deltagare DESC
LIMIT 5
```

```python
%sql
-- Vilket event har flest deltagare?
SELECT e.event_name, COUNT(DISTINCT f.athlete_id) as antal_deltagare
FROM marathos_catalog.marathon_gold.fct_results f
JOIN marathos_catalog.marathon_gold.dim_event e ON f.event_id = e.event_id
GROUP BY e.event_name
ORDER BY antal_deltagare DESC
LIMIT 5
```

```python

%sql
-- Visa de 5 vanligaste länderna bland löpare
SELECT athlete_country, COUNT(DISTINCT athlete_id) as antal_lopare
FROM marathos_catalog.marathon_gold.dim_athlete
WHERE athlete_country NOT LIKE 'XXX'
GROUP BY athlete_country
ORDER BY antal_lopare DESC
LIMIT 5
```
