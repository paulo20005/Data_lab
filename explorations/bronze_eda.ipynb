```python
df = spark.read.format("csv") \
  .option("header", "true") \
  .option("inferSchema", "true") \
  .load("/Volumes/marathos_catalog/marathon_raw/marathon_volume/TWO_CENTURIES_OF_UM_RACES.csv")

# Visa första 20 raderna
display(df.limit(20))
```

```python
# Antal rader
print(f"Antal rader: {df.count()}")

# Antal kolumner
print(f"Antal kolumner: {len(df.columns)}")
```

```python
# Visa ALLA kolumnnamn
print(df.columns)
```

```python
# Visa schema
df.printSchema()
```

```python
from pyspark.sql.functions import col, sum as spark_sum

# Räkna null-värden per kolumn
null_counts = df.select([spark_sum(col(c).isNull().cast("int")).alias(c) for c in df.columns])
display(null_counts)
```

```python
from pyspark.sql.functions import col, when, year, to_date

# Beräkna ålder = event_year - birth_year
df_with_age = df.withColumn("age", col("Year of event") - col("Athlete year of birth"))

# Visa åldersfördelning
display(df_with_age.select("age"))
```

```python
df.groupBy("Athlete country").count().orderBy("count", ascending=False).display()
```
